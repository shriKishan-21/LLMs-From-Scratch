# Import trained model


```python
# we'll use word2vec google news 300 model that is trained on goole news
# here 300 means the number of dimensions - each word is transformed into a 300 dimensional vector space
```

```python
import gensim.downloader as api
model = api.load("word2vec-google-news-300")
```

## Example of a word as a vector

```python
word_vectors = model

# Let us look how the vector embedding of a word looks like
print(word_vectors['computer']) # Example : Accessing the vector for the word 'computer'
```

```python
print(word_vectors['cat'].shape)
```

## King + Women - Men = 

```python
print(word_vectors.most_similar(positive=['king','women'],negative=['men'],topn=10))
```

## Let us check the similarity b/w a few pair of words

```python
# Example of calculating similarity
print(word_vectors.similarity('woman', 'man'))
print(word_vectors.similarity('king', 'queen'))
print(word_vectors.similarity('uncle', 'aunt'))
print(word_vectors.similarity('boy', 'girl'))
print(word_vectors.similarity('nephew', 'niece'))
print(word_vectors.similarity('paper', 'water'))
```

## Most similar words

```python
print(word_vectors.most_similar('house',topn=5))
```

## Now let us see the vector similarity

```python
import numpy as np
# Words to compare
word1 = 'man'
word2 = 'woman'

word3 = 'semiconductor'
word4 = 'earthworm'

word5 = 'nephew'
word6 = 'niece'

# calculate the vector difference
vector_differnec1 = word_vectors[word1] - word_vectors[word2]
vector_differnec2 = word_vectors[word3] - word_vectors[word4]
vector_differnec3 = word_vectors[word5] - word_vectors[word6]

# change the magnitude of the vector difference
magnitude_of_difference1 = np.linalg.norm(vector_differnec1)
magnitude_of_difference2 = np.linalg.norm(vector_differnec2)
magnitude_of_difference3 = np.linalg.norm(vector_differnec3)

# print the magnitude of the difference
print("The magnitude of the differnce between '{}' and '{}' is {:.2f}".format(word1,word2,magnitude_of_difference1))
print("The magnitude of the differnce between '{}' and '{}' is {:.2f}".format(word3,word4,magnitude_of_difference2))
print("The magnitude of the differnce between '{}' and '{}' is {:.2f}".format(word5,word6,magnitude_of_difference3))
```

```python
# Here the magnitude tell's us how far or how close the words are
# man-woman = 1.73 that mean they are close to each other
# semiconductor-earthworm = 5.67 they are far from each other
```
